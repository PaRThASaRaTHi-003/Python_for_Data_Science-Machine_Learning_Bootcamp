Let's explore the data a bit! Before continuing, I encourage you to check out the documentation on [Multi-Level Indexing](http://pandas.pydata.org/pandas-docs/stable/advanced.html) and [Using .xs](http://pandas.pydata.org/pandas-docs/stable/generated/pandas.DataFrame.xs.html).

**What is the max Close price for each bank's stock throughout the time period?**

```python
tickers = bank_stocks.columns.tolist()
tickers = list(sorted(set([x[0] for x in tickers])))
tickers
```

```python
for tick in tickers:
    print(tick, bank_stocks[tick]['Close'].max())
```

In one line clean code: use .xs()

```python
bank_stocks.xs(key='Close',axis=1,level='Stock Info').max()
```

**Create a new empty DataFrame called returns. This dataframe will contain the returns for each bank's stock. returns are typically defined by:**

$$r_t = \frac{p_t - p_{t-1}}{p_{t-1}} = \frac{p_t}{p_{t-1}} - 1$$

```python
returns = pd.DataFrame()
```

**We can use pandas pct_change() method on the Close column to create a column representing this return value. Create a for loop that goes and for each Bank Stock Ticker creates this returns column and set's it as a column in the returns DataFrame.**

```python
for tick in tickers:
    returns[tick+' Return'] = bank_stocks[tick]['Close'].pct_change()
returns.head()
```

## Moving Averages

Let's analyze the moving averages for these stocks in the year 2008.

**Plot the rolling 30 day average against the Close Price for Bank Of America's stock for the year 2008**

```python
BAC = bank_stocks['BAC']
BAC.head()
```

```python
plt.figure(figsize=(12,6))
BAC['Close'].loc['2008-01-01':'2009-01-01'].rolling(window = 30).mean().plot(label = 'BAC 30 Day Mov Avg')
BAC['Close'].loc['2008-01-01':'2009-01-01'].plot(label = 'BAC Close Price')
plt.legend()
```

**Create a heatmap of the correlation between the stocks Close Price.**

```python
bank_stocks.xs(key='Close',axis=1,level='Stock Info').corr()
```

This shows the correlation between Close prices of every stocks.

Now the heatmap between this relation

```python
sns.heatmap(bank_stocks.xs(key='Close',axis=1,level='Stock Info').corr(), annot=True)
```

**Optional: Use seaborn's clustermap to cluster the correlations together:**

```python
sns.clustermap(bank_stocks.xs(key='Close',axis=1,level='Stock Info').corr(), annot=True)
```
