```python
import pandas as pd
```

```python
sal = pd.read_csv('Salaries.csv', on_bad_lines='skip')
```

Check the head of the dataframe

```python
sal.head()
```

```python
sal.info()
```

average base-pay determination

```python
sal['BasePay'] = pd.to_numeric(sal['BasePay'], errors='coerce')
sal['BasePay'].mean()
```

highest overtime pay

```python
sal['OvertimePay'] = pd.to_numeric(sal['OvertimePay'], errors='coerce')
sal['OvertimePay'].max()
```

What is the Job title of JOSEPH DRISCOLL?

```python
sal[sal['EmployeeName']=='JOSEPH DRISCOLL']['JobTitle']
```

How much does JOSEPH DRISCOLL make including benefits?


```python
sal[sal['EmployeeName']=='JOSEPH DRISCOLL']['TotalPayBenefits']
```

Highest paid person finding:

```python
sal[sal['TotalPayBenefits']==sal['TotalPayBenefits'].max()]['EmployeeName']
```

Lowest paid person finding:

```python
sal[sal['TotalPayBenefits']==sal['TotalPayBenefits'].min()]['EmployeeName']
```

Average salary yearwise

```python
sal['Year'] = pd.to_numeric(sal['Year'], errors='coerce')
sal.dropna(subset=['Year'], inplace=True)
sal['Year'] = sal['Year'].astype(int)
sal.groupby('Year')['BasePay'].mean()
```

How many unique job titles are there?

```python
sal['JobTitle'].nunique()
```

Top 5 common jobs:

```python
sal['JobTitle'].value_counts().head(5)
```

Job titles with only one occurance in 2013

```python
sum(sal[sal['Year']==2013]['JobTitle'].value_counts()==1)
```

**How many people have Chief in their job title?**

```python
def chief_string(title):
    if 'chief' in title.lower().split():
        return True
    else:
        return False
```

```python
sum(sal['JobTitle'].apply(lambda x: chief_string(x)))
```
